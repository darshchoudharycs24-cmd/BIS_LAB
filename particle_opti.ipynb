```python
import random

def f(x):
    """Objective function: Minimize f(x) = x^2"""
    return x * x

def main():
    n = 4  # Number of particles
    
    # Initialization
    x = [random.uniform(0, 10) for _ in range(n)]  # Particle positions
    v = [random.random() for _ in range(n)]         # Particle velocities
    p = list(x)                                    # Personal best positions
    
    # Set initial global best to the first particle's position
    g = p[0]
    for i in range(1, n):
        if f(p[i]) < f(g):
            g = p[i]

    # Optimization Loop (10 iterations)
    for t in range(10):
        for i in range(n):
            # Update velocity & position
            v[i] = 0.5 * v[i] + 1.5 * random.random() * (p[i] - x[i]) + 1.5 * random.random() * (g - x[i])
            x[i] += v[i]

            # Update personal best
            if f(x[i]) < f(p[i]):
                p[i] = x[i]
            
            # Update global best
            if f(p[i]) < f(g):
                g = p[i]

    # Output results
    print(f"Best position: {g}")
    print(f"Best fitness: {f(g)}")

if __name__ == "__main__":
    main()
```
