```python
import random

cities = ['A', 'B', 'C', 'D', 'E']
distances = {
    ('A', 'B'): 10, ('A', 'C'): 15, ('A', 'D'): 20, ('A', 'E'): 25,
    ('B', 'C'): 35, ('B', 'D'): 25, ('B', 'E'): 30,
    ('C', 'D'): 30, ('C', 'E'): 20,
    ('D', 'E'): 15
}

def get_dist(c1, c2):
    return distances.get((c1, c2)) or distances.get((c2, c1)) or 0

def total_distance(route):
    dist = sum(get_dist(route[i], route[i+1]) for i in range(len(route)-1))
    dist += get_dist(route[-1], route[0])
    return dist

def create_route():
    route = list(cities)
    random.shuffle(route)
    return route

def mutate(route):
    i, j = random.sample(range(len(cities)), 2)
    route[i], route[j] = route[j], route[i]
    return route

population = [create_route() for _ in range(10)]

for generation in range(20):
    population.sort(key=total_distance)
    next_generation = population[:2]
    
    while len(next_generation) < 10:
        parent = random.choice(population[:5])
        child = mutate(list(parent))
        next_generation.append(child)
        
    population = next_generation

best_route = population[0]
print("Best Route:", " -> ".join(best_route) + f" -> {best_route[0]}")
print("Total Distance:", total_distance(best_route))
```
